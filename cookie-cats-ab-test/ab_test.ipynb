```python
import pandas as pd
from scipy import stats

df = pd.read_csv('cookie_cats.csv')
df.shape
df.head()
df.describe()
```

```python
# 游戏局数有个 49854 的极端值，离谱
# 先 log 变换看看
```

```python
df['version'].value_counts()

# gate_30 44700, gate_40 45489，差一点，跑个卡方看看
obs = df.version.value_counts().reindex(['gate_30','gate_40']).values
print(stats.chisquare(obs, f_exp=[len(df)/2, len(df)/2]))
```

```python
def rate(group):
    return group.mean()

df.groupby('version')[['retention_1','retention_7']].mean()
```

```python
# 分别算两组留存
x = df[df.version=='gate_30'].retention_7.sum()
n30 = len(df[df.version=='gate_30'])
x40 = df[df.version=='gate_40'].retention_7.sum()
n40 = len(df[df.version=='gate_40'])
p = (x+x40)/(n30+n40)
se = (p*(1-p)*(1/n30+1/n40))**0.5
z = (x/n30 - x40/n40)/se
print(z, 2*(1-stats.norm.cdf(abs(z))))
```

```python
import numpy as np
q = df.sum_gamerounds.quantile(.99)
g30 = df[df.version=='gate_30'].sum_gamerounds
g40 = df[df.version=='gate_40'].sum_gamerounds
print(stats.ttest_ind(np.log1p(g30), np.log1p(g40), equal_var=False))
print(stats.ttest_ind(g30[g30<=q], g40[g40<=q], equal_var=False))
print(stats.mannwhitneyu(g30, g40, alternative='two-sided'))
```
